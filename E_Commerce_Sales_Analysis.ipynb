```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn



reviews = pd.read_csv('/content/olist_order_review.csv')
items = pd.read_csv('/content/olist_order_items.csv')
orders = pd.read_csv('/content/olist_orders_data.csv')
products = pd.read_csv('/content/olist_products_data.csv')
```

```python
orders.isnull().sum()
products.isnull().sum()
reviews.isnull().sum()
```

**Orders data cleaning **


```python
orders.isnull().sum()
```

```python
orders.drop_duplicates(inplace=True)
```

```python
orders['order_purchase_timestamp'] = pd.to_datetime(
orders['order_purchase_timestamp']
)
```

```python
orders.head()
```

**Products Data Cleaning**

```python
products.isnull().sum()
```

```python
products.drop_duplicates(inplace=True)
```

```python
products.head()
```

**Review Data Cleaning**

```python
reviews.isnull().sum()
```

```python
products.drop_duplicates(inplace=True)
```

```python
products.head()
```

```python
merged_dataset = orders.merge(
items,
on='order_id'
)

merged_dataset = merged_dataset.merge(
products,
on='product_id'
)

merged_dataset = merged_dataset.merge(
reviews,
on='order_id'
)
```

```python
merged_dataset.to_csv(
'final_dataset2.csv',
index=False
)
```

```python
merged_dataset.head()
```

**Highest Revenue Category**

```python
merged_dataset.groupby(
'product_category_name'
)['price'].sum().sort_values(ascending=False)
```

**Peak Sales Month**

```python
merged_dataset['Month'] = merged_dataset[
'order_purchase_timestamp'
].dt.month_name()

merged_dataset.groupby(
'Month'
)['price'].sum()
```

**Average Order Value**

```python
avg_Order = merged_dataset.groupby(
'Month'
)['price'].mean()
```

```python
merged_dataset.head(10)
```

```python
merged_dataset.to_csv(
'final_dataset2.csv',
index=False
)
```

**Visualisations**

```python
merged_dataset['review_score'].value_counts()
```

Chart 1: Revenue by Category

```python
top_cat = merged_dataset.groupby(
'product_category_name'
)['price'].sum().sort_values(
ascending=False
).head(10)

top_cat.plot(
kind='bar',
figsize=(12,6)
)

plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue')

plt.xticks(rotation=45, ha='right')

plt.tight_layout()



plt.savefig('Revenue_by_Category.png')
plt.show()
```

```python
from google.colab import files
files.download('Revenue_by_Category.png')
```

**Chart 2: Monthly Sales Trend**

```python
monthly = merged_dataset.groupby(
'Month'
)['price'].sum()

monthly.plot(
kind='line',
marker='o'
)

plt.title('Monthly Sales Trend')

plt.xlabel('Month')
plt.ylabel('Sales')

plt.savefig('Monthly_Sales_Trend.png')
plt.show()
```

```python
from google.colab import files
files.download('Monthly_Sales_Trend.png')
```

**Chart 3: Review Score Distribution**

```python
reviews['review_score'].value_counts().plot(
kind='pie',
autopct='%1.1f%%'
)

plt.title(
'Review Score Distribution'
)

plt.savefig('Review_Score_Distribution.png')

plt.show()
```

```python
from google.colab import files
files.download("Review_Score_Distribution.png")
```

**Chart 4: Order Value Distribution**

```python
plt.figure(figsize=(10,5))

plt.hist(
merged_dataset['price'],
bins=30
)

plt.title(
'Order Value Distribution'
)

plt.xlabel('Price')
plt.ylabel('Frequency')

plt.savefig('Order_value_dist.png')

plt.show()
```

```python
from google.colab import files
files.download('Order_value_dist.png')
```

**Chart 5: Category vs Month Heatmap**

```python
pivot = pd.pivot_table(
merged_dataset,
values='price',
index='product_category_name',
columns='Month',
aggfunc='sum'
)

plt.figure(figsize=(12,8))

sn.heatmap(
pivot,
cmap='YlGnBu'
)

plt.title(
'Category vs Month Heatmap'
)

plt.savefig('Category_vs_Month_Heatmap.png')

plt.show(10)
```

```python
from google.colab import files
files.download('Category_vs_Month_Heatmap.png')
```

**Chart 6: Top 10 Categories**

```python
top_cat = merged_dataset.groupby(
'product_category_name'
)['order_id'].count()

top_cat = top_cat.sort_values(
ascending=False
).head(10)

top_cat.plot(
kind='barh',
figsize=(10,6)
)

plt.title(
'Top 10 Categories by Orders'
)

plt.xlabel('Number of Orders')
plt.ylabel('Category')

plt.savefig('Top_10_Categories.png')

plt.show()
```

```python
from google.colab import files
files.download('Top_10_Categories.png')
```

```python
status_category = pd.crosstab(
    merged_dataset['product_category_name'],
    merged_dataset['order_status']
)

status_category.head(10).plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Order Status by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig('Order_Status_by_Category.png')

plt.show()
```
